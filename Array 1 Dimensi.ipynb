```python
#menyertakan modul numpy
import numpy as np

arr = np.array([10,30,50,60,90])
print("Daftar Angka :")
print(arr)
```

Aritmatika

```python
nilai_tertinggi = np.max(arr)
print(nilai_tertinggi)
```

```python
nilai_terendah = np.min(arr)
print(nilai_terendah)
```

```python
nilai_rata = np.mean(arr)
print(nilai_rata)
```

```python
nilai_tengah = np.median(arr)
print(nilai_tengah)
```
